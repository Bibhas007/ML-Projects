# **Example of pyspark ML**
predict final exam score of students

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('ML').getOrCreate()
```

```python
training = spark.read.csv(path,header=True,inferSchema=True)
training.show(2)
```

```python
training.printSchema()
```

```python
training.columns
```

```python
from pyspark.ml.feature import VectorAssembler
featureassembler = VectorAssembler(inputCols=['study_time_hours','attendance_percent','sleep_hours','previous_grade','final_exam_score'],outputCol='Independent Features')
```

```python
output =featureassembler.transform(training)
```

```python
output.show(5)
```

```python
pyspark_finalized_data = output.select('Independent Features','final_exam_score')
```

```python
pyspark_finalized_data.show(5)
```

```python
from pyspark.ml.regression import LinearRegression

# train-test split

train_data,test_data = pyspark_finalized_data.randomSplit([0.75,0.25])
```

```python
regressor = LinearRegression(featuresCol='Independent Features',labelCol='final_exam_score')
regressor = regressor.fit(train_data)
```

```python
regressor.coefficients
```

```python
regressor.intercept
```

```python
pred_results = regressor.evaluate(test_data)
```

```python
pred_results.predictions.show()
```

```python
pred_results.meanAbsoluteError,pred_results.meanSquaredError
```
